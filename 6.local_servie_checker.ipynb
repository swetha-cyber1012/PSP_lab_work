# 🔌 Local Service Checker

## 🎯 Aim

To write a Python program using the **Socket Network Utility** to check whether common services running on the local computer (`localhost`) are available on selected ports.

## 📝 Problem Statement

A developer may run different services on a local computer, such as a development web server, database server, or testing server. This program checks whether a selected **localhost port** is open and accepting connections.

The program uses Python's `socket` module to connect to the selected port and reports whether the service is **Running** or **Not Running**.

## ✨ Features

1. 🌐 Check common localhost ports.
2. 🔌 Test whether a selected port is open.
3. ✅ Display whether the service is running.
4. 🏷️ Identify common services based on their port numbers.
5. 🔄 Check multiple ports.
6. 📋 Display a simple service-check history.
7. ⚠️ Handle connection errors and invalid port numbers.
8. 🚪 Provide an option to exit the program.

## 🔢 Common Localhost Ports

| Port | Common Service |
|---:|---|
| 80 | HTTP Web Server |
| 443 | HTTPS Web Server |
| 3000 | Node.js / React Development Server |
| 5000 | Flask / Python Development Server |
| 8000 | Python Development Server |
| 8080 | Web Development / Testing Server |
| 3306 | MySQL |
| 5432 | PostgreSQL |

## 🔄 Working Process

```text
Enter Localhost Port
        ↓
Create Socket
        ↓
Try to Connect
        ↓
Connection Successful?
     ↙          ↘
   YES           NO
    ↓             ↓
🟢 Running     🔴 Not Running

````python
# 🔌 LOCAL SERVICE CHECKER
# Check services running on localhost

import socket

# Common localhost services
services = {
    80: "HTTP Web Server",
    443: "HTTPS Web Server",
    3000: "Node.js / React Development Server",
    5000: "Flask / Python Development Server",
    8000: "Python Development Server",
    8080: "Web Development / Testing Server",
    3306: "MySQL",
    5432: "PostgreSQL"
}


# Function to check a port
def check_port(port):

    service = services.get(port, "Unknown / Custom Service")

    print("\n" + "=" * 55)
    print("🔌 LOCAL SERVICE CHECKER")
    print("=" * 55)

    print("Host    : localhost")
    print("Port    :", port)
    print("Service :", service)

    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(1)

    try:
        # Try to connect to localhost
        result = sock.connect_ex(("127.0.0.1", port))

        if result == 0:
            print("Status  : 🟢 RUNNING")
        else:
            print("Status  : 🔴 NOT RUNNING")

    except socket.error as error:
        print("Status  : ⚠️ ERROR")
        print("Error   :", error)

    finally:
        sock.close()

    print("=" * 55)


# Function to display common ports
def show_ports():

    print("\n" + "=" * 60)
    print("📋 COMMON LOCALHOST PORTS")
    print("=" * 60)

    for port, service in services.items():
        print(f"{port:<8} → {service}")

    print("=" * 60)


# Main program
while True:

    print("\n" + "=" * 55)
    print("🔌 LOCAL SERVICE CHECKER")
    print("=" * 55)

    print("1. 🔍 Check a Port")
    print("2. 📋 Show Common Ports")
    print("3. 🚪 Exit")

    print("=" * 55)

    choice = input("Enter your choice (1-3): ").strip()

    if choice == "1":

        try:
            port = int(input("\nEnter localhost port number: "))

            if 1 <= port <= 65535:
                check_port(port)
            else:
                print("❌ Port must be between 1 and 65535.")

        except ValueError:
            print("❌ Please enter a valid number.")

    elif choice == "2":

        show_ports()

    elif choice == "3":

        print("\n👋 Thank you for using Local Service Checker!")
        break

    else:

        print("\n❌ Invalid choice!")
        print("Please select 1, 2, or 3.")
````
