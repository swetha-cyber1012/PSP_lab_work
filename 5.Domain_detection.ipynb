# 🔎 Domain Detective – Network Utility

## 🎯 Aim

To write a Python program using the **Network Utility `socket` module** to find the IP address of a given domain name such as `example.com`.

## 📝 Problem Statement

A domain name is easy for humans to remember, but computers communicate using IP addresses. This program accepts a domain name from the user and uses Python's built-in **socket network utility** to find and display its corresponding IP address.

## ✨ Features

1. 🌐 Enter any permitted domain name.
2. 🔎 Find the IPv4 address of the domain.
3. 📋 Display the domain and IP address clearly.
4. 🔄 Check multiple domains.
5. ⚠️ Handle invalid or unavailable domains.
6. 📊 Maintain a simple lookup history.
7. 🚪 Exit the program safely.

## 🧰 Network Utility Used

```python

````python
# 🔎 DOMAIN DETECTIVE
# Find IP Address using Network Utility

import socket

while True:
    print("\n" + "=" * 45)
    print("🔎 DOMAIN DETECTIVE")
    print("=" * 45)

    print("1. Find IP Address")
    print("2. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        domain = input("\nEnter domain name: ").strip()

        try:
            # Network utility for DNS lookup
            ip_address = socket.gethostbyname(domain)

            print("\n✅ DOMAIN FOUND")
            print("-" * 35)
            print("Domain     :", domain)
            print("IP Address :", ip_address)
            print("-" * 35)

        except socket.gaierror:
            print("\n❌ Domain not found!")
            print("Please check the domain name.")

    elif choice == "2":

        print("\n👋 Thank you for using Domain Detective!")
        break

    else:

        print("\n❌ Invalid choice! Please select 1 or 2.")
````
