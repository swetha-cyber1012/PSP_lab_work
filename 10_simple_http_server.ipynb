# 10. Digital College Notice Board

**Topic:** HTTP Server

## Aim
Serve a simple local notice board page at http://127.0.0.1:8000. Stop the server with Ctrl+C.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **HTTP Server**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
from http.server import BaseHTTPRequestHandler, HTTPServer
from html import escape

NOTICES = [
    ("15 Oct", "Internal assessment timetable published"),
    ("18 Oct", "Coding club meeting at 3:00 PM"),
    ("22 Oct", "Library books due for return"),
]

class NoticeHandler(BaseHTTPRequestHandler):
    def do_GET(self):
        items = "".join(
            f"<li><strong>{escape(day)}:</strong> {escape(notice)}</li>"
            for day, notice in NOTICES
        )
        page = f"""<!doctype html>
<html><head><meta charset="utf-8"><title>College Notice Board</title></head>
<body><h1>Digital College Notice Board</h1><ul>{items}</ul></body></html>"""
        body = page.encode("utf-8")
        self.send_response(200)
        self.send_header("Content-Type", "text/html; charset=utf-8")
        self.send_header("Content-Length", str(len(body)))
        self.end_headers()
        self.wfile.write(body)

    def log_message(self, format, *args):
        print("%s - %s" % (self.address_string(), format % args))

server = HTTPServer(("127.0.0.1", 8000), NoticeHandler)
print("Open http://127.0.0.1:8000 in your browser. Press Ctrl+C to stop.")
try:
    server.serve_forever()
except KeyboardInterrupt:
    print("\\nServer stopped.")
finally:
    server.server_close()
```
