# 20. Student Club Registration

**Topic:** Secure Input Handling

## Aim
Validate name, email, phone number, and password before saving non-password registration details.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **Secure Input Handling**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
import re
import json
from pathlib import Path
from getpass import getpass

name = input("Full name: ").strip()
email = input("Email: ").strip()
phone = input("Phone number (10 digits): ").strip()
password = getpass("Password (min 8 chars, upper/lower/digit): ")

errors = []
if len(name) < 2:
    errors.append("Name must contain at least 2 characters.")
if not re.fullmatch(r"[^@\s]+@[^@\s]+\.[^@\s]+", email):
    errors.append("Enter a valid email address.")
if not re.fullmatch(r"\d{10}", phone):
    errors.append("Phone number must contain exactly 10 digits.")
if not (len(password) >= 8 and re.search(r"[A-Z]", password)
        and re.search(r"[a-z]", password) and re.search(r"\d", password)):
    errors.append("Password must be 8+ characters with uppercase, lowercase, and a digit.")

if errors:
    print("Registration not saved:")
    for error in errors:
        print("-", error)
else:
    # Never save the plaintext password. This demo stores only contact fields.
    file_path = Path("club_registrations.json")
    registrations = []
    if file_path.exists():
        try:
            registrations = json.loads(file_path.read_text(encoding="utf-8"))
        except json.JSONDecodeError:
            print("Existing registration file is invalid; starting a new list.")
    registrations.append({"name": name, "email": email, "phone": phone})
    file_path.write_text(json.dumps(registrations, indent=2), encoding="utf-8")
    print("Registration saved. Password was not stored.")
```
