**Topic:** Host Discovery

## Aim
Check only a predefined, explicitly authorized private lab range. Set LAB_NETWORK to your approved range before running.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **Host Discovery**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
import ipaddress
import socket

# IMPORTANT: Change this to a small network you own or have permission to test.
LAB_NETWORK = "192.168.1.0/29"  # Small example range; edit for your authorized lab.
TIMEOUT_SECONDS = 0.25

network = ipaddress.ip_network(LAB_NETWORK, strict=False)
if not network.is_private:
    raise ValueError("For safety, configure a private lab network only.")
print(f"Checking {network} (authorized lab use only)")
for address in network.hosts():
    try:
        # Reverse DNS is a gentle, non-scanning check; it may not identify every live host.
        name, aliases, _ = socket.gethostbyaddr(str(address))
        print(f"{address} responds to reverse lookup: {name}")
    except (socket.herror, socket.gaierror, OSError):
        pass
print("Finished. Missing results do not mean a device is offline.")
```
