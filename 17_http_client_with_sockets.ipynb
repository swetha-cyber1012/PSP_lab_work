# 17. Mini Web Browser

**Topic:** HTTP Request Using Socket

## Aim
Fetch a basic page from a permitted plain-HTTP test host using a raw TCP socket.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **HTTP Request Using Socket**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
import socket

host = input("Authorized HTTP test host (default: example.com): ").strip() or "example.com"
path = "/"
request = (
    f"GET {path} HTTP/1.1\\r\\n"
    f"Host: {host}\\r\\n"
    "User-Agent: CollegeSocketBrowser/1.0\\r\\n"
    "Connection: close\\r\\n\\r\\n"
)
try:
    with socket.create_connection((host, 80), timeout=5) as sock:
        sock.sendall(request.encode("ascii"))
        chunks = []
        while True:
            data = sock.recv(4096)
            if not data:
                break
            chunks.append(data)
    response = b"".join(chunks).decode("utf-8", errors="replace")
    print(response[:8000])
except (OSError, UnicodeError) as exc:
    print("Request failed:", exc)
```
