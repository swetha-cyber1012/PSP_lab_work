# 14. College Portal Login

**Topic:** Input Validation

## Aim
Validate a student ID and password input and display helpful error messages. This is a classroom demo, not real authentication.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **Input Validation**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
import re
import getpass

student_id = input("Student ID (e.g. STU1234): ").strip()
password = getpass.getpass("Password: ")

errors = []
if not student_id:
    errors.append("Student ID cannot be empty.")
elif not re.fullmatch(r"STU\d{4}", student_id, flags=re.IGNORECASE):
    errors.append("Student ID must look like STU1234.")
if not password:
    errors.append("Password cannot be empty.")
elif len(password) < 8:
    errors.append("Password must contain at least 8 characters.")

if errors:
    print("Validation failed:")
    for error in errors:
        print("-", error)
else:
    print("Input format is valid. No real login was performed.")
```
