# 💼 Employee Salary Analyzer
### CSV-Based Employee Salary Analysis System

## 🎯 Aim

To develop a Python program that reads employee details from a CSV file and displays employees whose salary is above a user-defined salary limit.

## 📌 Problem Statement

An organization stores employee information such as Employee ID, Name, Department, and Salary in a CSV file.

The program should:
- Read employee data from a CSV file.
- Ask the user to enter a minimum salary.
- Find employees earning above that salary.
- Display the matching employee details.
- Calculate the number of employees and average salary of the selected employees.

## ⭐ Features

- 📂 Read employee data from CSV
- 💰 Enter user-defined salary limit
- 🔍 Find employees earning above the limit
- 📋 Display employee details
- 👥 Count matching employees
- 📊 Calculate average salary
- ⚠️ Handle invalid input and file errors

## 📄 CSV File Format

The CSV file contains:

| Employee ID | Name | Department | Salary |
|---|---|---|---:|
| E101 | Anu | HR | 30000 |
| E102 | Ravi | IT | 45000 |
| E103 | Priya | Finance | 55000 |
| E104 | Karthik | IT | 60000 |
| E105 | Meena | HR | 35000 |

## 🔄 Working

1. Create or read the employee CSV file.
2. Read employee records using Python's `csv` module.
3. Ask the user to enter a salary limit.
4. Compare each employee's salary with the limit.
5. Display employees whose salary is greater than the limit.
6. Display the total number of matching employees.
7. Calculate their average salary.

## 🧠 Concepts Used

- CSV File Handling
- `csv` module
- Lists
- Dictionaries
- Functions
- `for` loop
- `if` condition
- User input
- Exception handling
- Arithmetic operations

## 🏁 Conclusion

The Employee Salary Analyzer demonstrates how Python can read employee information from a CSV file and perform simple salary-based data analysis.

```python
# 💼 EMPLOYEE SALARY ANALYZER
# Read employee data from CSV
# Display employees earning above a user-defined salary

import csv
import os

FILE_NAME = "employees.csv"


# 📄 Create sample CSV file
def create_csv_file():
    if not os.path.exists(FILE_NAME):
        employees = [
            ["Employee ID", "Name", "Department", "Salary"],
            ["E101", "Anu", "HR", 30000],
            ["E102", "Ravi", "IT", 45000],
            ["E103", "Priya", "Finance", 55000],
            ["E104", "Karthik", "IT", 60000],
            ["E105", "Meena", "HR", 35000],
            ["E106", "Arun", "Cyber Security", 65000],
            ["E107", "Divya", "IT", 50000]
        ]

        with open(FILE_NAME, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerows(employees)

        print("✅ Sample employee CSV file created.")


# 📖 Read employee data
def read_employees():
    employees = []

    try:
        with open(FILE_NAME, "r", newline="") as file:
            reader = csv.DictReader(file)

            for row in reader:
                row["Salary"] = float(row["Salary"])
                employees.append(row)

        return employees

    except FileNotFoundError:
        print("❌ CSV file not found.")
        return []

    except ValueError:
        print("❌ Invalid salary value in CSV file.")
        return []


# 💰 Analyze salary
def analyze_salary():
    employees = read_employees()

    if not employees:
        print("❌ No employee data available.")
        return

    try:
        minimum_salary = float(
            input("\n💰 Enter minimum salary: ")
        )

        matching_employees = []

        for employee in employees:
            if employee["Salary"] > minimum_salary:
                matching_employees.append(employee)

        print("\n" + "=" * 75)
        print("💼 EMPLOYEES EARNING ABOVE SALARY LIMIT")
        print("=" * 75)

        if not matching_employees:
            print("❌ No employees found above this salary.")
            return

        print(
            f"{'ID':<10}"
            f"{'Name':<15}"
            f"{'Department':<20}"
            f"{'Salary':>12}"
        )

        print("-" * 75)

        total_salary = 0

        for employee in matching_employees:
            print(
                f"{employee['Employee ID']:<10}"
                f"{employee['Name']:<15}"
                f"{employee['Department']:<20}"
                f"₹{employee['Salary']:>10,.2f}"
            )

            total_salary += employee["Salary"]

        average_salary = total_salary / len(matching_employees)

        print("-" * 75)
        print("👥 Employees Found :", len(matching_employees))
        print(f"📊 Average Salary  : ₹{average_salary:,.2f}")

    except ValueError:
        print("❌ Please enter a valid salary amount.")


# 📋 Display all employees
def display_all():
    employees = read_employees()

    if not employees:
        return

    print("\n" + "=" * 75)
    print("📋 ALL EMPLOYEE RECORDS")
    print("=" * 75)

    for employee in employees:
        print(
            f"{employee['Employee ID']} | "
            f"{employee['Name']} | "
            f"{employee['Department']} | "
            f"₹{employee['Salary']:,.2f}"
        )


# 🚀 Create CSV file first
create_csv_file()


# 🎯 Main Menu
while True:

    print("\n" + "=" * 55)
    print("💼 EMPLOYEE SALARY ANALYZER")
    print("=" * 55)

    print("1. 📋 Display All Employees")
    print("2. 💰 Analyze Salary")
    print("3. 🚪 Exit")

    print("=" * 55)

    choice = input("Enter your choice (1-3): ").strip()

    if choice == "1":
        display_all()

    elif choice == "2":
        analyze_salary()

    elif choice == "3":
        print("\n👋 Thank you for using Employee Salary Analyzer!")
        break

    else:
        print("❌ Invalid choice! Please select 1-3.")
```
