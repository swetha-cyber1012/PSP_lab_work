# 18. Smart ATM Simulator

**Topic:** Exception Handling

## Aim
Simulate an ATM with PIN validation, balance checks, and exception handling.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **Exception Handling**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
balance = 5000.0
correct_pin = "1234"  # Demo only; never store real PINs this way.

try:
    pin = input("Enter 4-digit demo PIN: ").strip()
    if pin != correct_pin:
        raise PermissionError("Incorrect PIN.")
    amount_text = input(f"Balance ₹{balance:.2f}. Amount to withdraw: ").strip()
    amount = float(amount_text)
    if amount <= 0:
        raise ValueError("Amount must be greater than zero.")
    if amount > balance:
        raise ArithmeticError("Insufficient balance.")
    balance -= amount
    print(f"Please collect ₹{amount:.2f}. Remaining balance: ₹{balance:.2f}")
except PermissionError as exc:
    print("Authentication error:", exc)
except ValueError as exc:
    print("Invalid amount:", exc)
except ArithmeticError as exc:
    print("Transaction declined:", exc)
```
