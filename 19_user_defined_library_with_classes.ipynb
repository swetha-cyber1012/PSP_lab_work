# 19. Digital Library System

**Topic:** User-Defined Module

## Aim
Create a reusable library module with Book and Library classes, then issue, return, and search books.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **User-Defined Module**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
from dataclasses import dataclass

@dataclass
class Book:
    book_id: str
    title: str
    author: str
    issued: bool = False

class Library:
    def __init__(self, books=None):
        self.books = {book.book_id: book for book in (books or [])}

    def search(self, keyword):
        keyword = keyword.casefold()
        return [b for b in self.books.values()
                if keyword in b.title.casefold() or keyword in b.author.casefold()]

    def issue(self, book_id):
        book = self.books.get(book_id)
        if not book:
            return "Book ID not found."
        if book.issued:
            return "Book is already issued."
        book.issued = True
        return f"Issued: {book.title}"

    def return_book(self, book_id):
        book = self.books.get(book_id)
        if not book:
            return "Book ID not found."
        if not book.issued:
            return "Book was not marked as issued."
        book.issued = False
        return f"Returned: {book.title}"

library = Library([
    Book("B01", "Python Basics", "A. Kumar"),
    Book("B02", "Computer Networks", "S. Rao"),
    Book("B03", "Data Structures", "M. Devi"),
])

print("Search results:")
for book in library.search(input("Search title/author: ").strip()):
    print(f"{book.book_id}: {book.title} — {book.author} | {'Issued' if book.issued else 'Available'}")
book_id = input("Book ID to issue: ").strip().upper()
print(library.issue(book_id))
if input("Return this book now? (y/n): ").strip().lower() == "y":
    print(library.return_book(book_id))
```
