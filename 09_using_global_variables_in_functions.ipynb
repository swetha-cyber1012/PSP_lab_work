# 09. Hostel Electricity Tracker

**Topic:** Global Variable

## Aim
Calculate electricity bills for rooms and keep a running total with a global variable.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **Global Variable**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
total_bill = 0.0  # Global total for this program run

def calculate_room_bill():
    global total_bill
    room = input("Room number: ").strip()
    try:
        previous = float(input("Previous meter reading (kWh): "))
        current = float(input("Current meter reading (kWh): "))
        rate = float(input("Cost per unit (₹): "))
        if not room or previous < 0 or current < previous or rate < 0:
            raise ValueError("Check room, readings, and rate.")
    except ValueError as exc:
        print("Invalid input:", exc)
        return
    units = current - previous
    bill = units * rate
    total_bill += bill
    print(f"Room {room}: {units:.2f} units, bill ₹{bill:.2f}")
    print(f"Running hostel total: ₹{total_bill:.2f}")

while input("Add room bill? (y/n): ").strip().lower() == "y":
    calculate_room_bill()
print(f"Final total bill: ₹{total_bill:.2f}")
```
