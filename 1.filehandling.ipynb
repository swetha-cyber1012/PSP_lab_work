# 🎓 Smart Weekly Attendance Diary

## Aim

To write a Python program using **File Handling** to store the weekly attendance
details of multiple students from Monday to Friday and calculate their weekly
attendance percentage.

## Working

The program records attendance for 5 working days:

- Monday
- Tuesday
- Wednesday
- Thursday
- Friday

Each day is worth **20%**.

### Formula

**Weekly Attendance Percentage = (Number of Present Days / 5) × 100**

### Example

| Day | Attendance |
|---|---|
| Monday | Present |
| Tuesday | Present |
| Wednesday | Absent |
| Thursday | Present |
| Friday | Absent |

Present Days = 3

Weekly Attendance = (3 / 5) × 100 = **60%**

## Features

- Add multiple students
- Record Monday to Friday attendance
- Mark Present or Absent
- Calculate weekly attendance percentage
- Store records in `attendance.txt`
- Read records using File Handling
- Display all students and their attendance percentage in a table

## File Handling Concepts Used

- `open()`
- Append mode (`a`)
- Read mode (`r`)
- `write()`
- `read()`
- `with open()`
- `FileNotFoundError`

```python
# 🎓 Smart Weekly Attendance Diary
# Multiple Students + File Handling

FILE_NAME = "attendance.txt"

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

students = []

# ------------------------------------------------
# Get Number of Students
# ------------------------------------------------

while True:
    try:
        n = int(input("Enter number of students: "))

        if n > 0:
            break
        else:
            print("❌ Please enter a number greater than 0.")

    except ValueError:
        print("❌ Please enter a valid number.")


# ------------------------------------------------
# Enter Attendance for Each Student
# ------------------------------------------------

for i in range(n):

    print("\n" + "=" * 45)
    print(f"👩‍🎓 STUDENT {i + 1}")
    print("=" * 45)

    name = input("Enter student name: ").strip()

    while name == "":
        print("❌ Student name cannot be empty.")
        name = input("Enter student name: ").strip()

    attendance = {}
    present_days = 0

    print("\nEnter P for Present or A for Absent")

    # Monday to Friday
    for day in days:

        while True:

            status = input(f"{day}: ").strip().upper()

            if status == "P" or status == "A":
                break

            print("❌ Invalid input! Please enter P or A.")

        attendance[day] = status

        if status == "P":
            present_days += 1

    # ------------------------------------------------
    # Calculate Weekly Attendance
    # ------------------------------------------------

    percentage = (present_days / 5) * 100

    # Store student information
    student = {
        "name": name,
        "attendance": attendance,
        "present_days": present_days,
        "absent_days": 5 - present_days,
        "percentage": percentage
    }

    students.append(student)


# ------------------------------------------------
# Save Records to File
# ------------------------------------------------

try:

    with open(FILE_NAME, "a") as file:

        for student in students:

            file.write(f"Student Name: {student['name']}\n")

            for day in days:

                if student["attendance"][day] == "P":
                    status = "Present"
                else:
                    status = "Absent"

                file.write(f"{day}: {status}\n")

            file.write(
                f"Present Days: {student['present_days']}/5\n"
            )

            file.write(
                f"Absent Days: {student['absent_days']}/5\n"
            )

            file.write(
                f"Weekly Attendance: {student['percentage']:.2f}%\n"
            )

            file.write("-" * 45 + "\n")

    print("\n✅ All attendance records saved successfully!")

except Exception as e:
    print("❌ Error while saving file:", e)


# ------------------------------------------------
# Display Weekly Attendance Summary
# ------------------------------------------------

print("\n")
print("=" * 55)
print("📊 WEEKLY ATTENDANCE SUMMARY")
print("=" * 55)

print(f"{'Student Name':<25}{'Attendance %':>15}")

print("-" * 55)

for student in students:

    print(
        f"{student['name']:<25}"
        f"{student['percentage']:>14.2f}%"
    )

print("=" * 55)


# ------------------------------------------------
# Read Saved Records From File
# ------------------------------------------------

print("\n📂 SAVED ATTENDANCE RECORDS")
print("=" * 55)

try:

    with open(FILE_NAME, "r") as file:

        records = file.read()

        if records.strip():
            print(records)
        else:
            print("No attendance records found.")

except FileNotFoundError:

    print("❌ Attendance file does not exist yet.")

except Exception as e:

    print("❌ Error while reading file:", e)
```
