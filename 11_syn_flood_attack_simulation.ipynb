# 11. Rate-Limit Security Lab

**Topic:** Security

## Aim
Demonstrate a simple rate limiter locally by accepting only a small number of attempts per time window.

## Algorithm
1. Import the required Python modules.
2. Accept or prepare the required input/data.
3. Process the data using the concept **Security**.
4. Display the result and handle common errors.

## Program
Run the code cell below. Some networking examples are limited to localhost or authorized test systems.


```python
import time
from collections import deque

MAX_REQUESTS = 3
WINDOW_SECONDS = 5
request_times = deque()

print(f"Demo: maximum {MAX_REQUESTS} requests per {WINDOW_SECONDS} seconds.")
print("This is a local simulation; no external server is contacted.")
for attempt in range(1, 7):
    now = time.monotonic()
    while request_times and now - request_times[0] >= WINDOW_SECONDS:
        request_times.popleft()
    if len(request_times) < MAX_REQUESTS:
        request_times.append(now)
        print(f"Request {attempt}: ACCEPTED")
    else:
        wait = max(0, WINDOW_SECONDS - (now - request_times[0]))
        print(f"Request {attempt}: RATE-LIMITED (try again in about {wait:.1f}s)")
    time.sleep(0.4)
```
