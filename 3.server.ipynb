# 💬 College Messenger – TCP Client-Server Communication System

## 🎯 Aim

To write a Python **TCP Client-Server program** that allows a college student to send messages such as **"Good Morning"**, **"Assignment Submitted"**, **"Class Started"**, or **"Need Help"** from the client to the server and receive a response from the server.

## 📝 Problem Statement

A college needs a simple messaging system for communication between students and a server. The **Client** sends a message to the **Server** using TCP socket communication. The Server receives the message, displays it, processes it, and sends a response back to the Client.

The system provides a simple menu-driven interface so that students can easily send different types of messages.

## ✨ Features

The College Messenger includes the following useful features:

1. 💬 **Send Message** – Send a custom message from the client to the server.
2. 👋 **Quick Messages** – Select commonly used messages such as:
   - Good Morning
   - Assignment Submitted
   - Class Started
   - Need Help
   - Thank You
3. 📢 **Message Category** – Select a category such as:
   - Academic
   - Attendance
   - Assignment
   - General
   - Emergency
4. 🕒 **Message Time** – Automatically record the time when a message is sent.
5. 📋 **Message History** – Store and display messages sent during the session.
6. ✅ **Server Acknowledgement** – The server confirms that the message was received successfully.
7. 🔢 **Message Counter** – Display the total number of messages sent.
8. 🚪 **Disconnect Option** – Safely close the client-server connection.
9. ⚠️ **Error Handling** – Handle connection errors and invalid input properly.

## 🔄 Working Process

The system works using **TCP Socket Programming**.

```text
        CLIENT
           │
           │  Connect
           ▼
        SERVER
           │
           │  Accept Connection
           ▼
     Send Message
           │
           ▼
     Server Receives
           │
           ▼
    Process Message
           │
           ▼
   Send Acknowledgement
           │
           ▼
        CLIENT

````python
# 💬 COLLEGE MESSENGER - SERVER

import socket
from datetime import datetime

HOST = "127.0.0.1"
PORT = 5000

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind((HOST, PORT))
server.listen(1)

print("=" * 55)
print("       🖥️ COLLEGE MESSENGER SERVER")
print("=" * 55)
print(f"Server started on {HOST}:{PORT}")
print("Waiting for client connection...\n")

conn, address = server.accept()

print(f"✅ Client connected: {address}")
print("-" * 55)

message_count = 0

while True:
    try:
        data = conn.recv(1024).decode()

        if not data:
            break

        if data == "EXIT":
            print("🚪 Client disconnected.")
            break

        message_count += 1

        # Split category and message
        parts = data.split("|", 1)

        category = parts[0]
        message = parts[1]

        current_time = datetime.now().strftime("%H:%M:%S")

        print(f"📨 Message {message_count}")
        print(f"🕒 Time     : {current_time}")
        print(f"📂 Category : {category}")
        print(f"💬 Message  : {message}")
        print("-" * 55)

        # Send acknowledgement
        response = (
            f"✅ Message received successfully!\n"
            f"📨 Message Number: {message_count}"
        )

        conn.send(response.encode())

    except ConnectionError:
        print("❌ Connection lost.")
        break

conn.close()
server.close()

print("\n🔴 Server stopped.")
````
