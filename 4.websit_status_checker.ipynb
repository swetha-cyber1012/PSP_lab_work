# 🌐 Website Health Monitor

## 🎯 Aim

To write a Python program that checks whether a permitted college or test website is **online and accessible**, and displays important information such as the **HTTP status code, response time, and website status**.

## 📝 Problem Statement

A college IT team needs a simple tool to monitor the availability of permitted websites used for academic and testing purposes. The program accepts a website URL, sends a request to the website, and checks whether the website is reachable.

The program displays the website's **HTTP status code, response time, and availability status** in a simple and user-friendly format.

## ✨ Features

1. 🌐 **Website Availability Check** – Check whether the website is online.
2. 🔢 **HTTP Status Code** – Display the response status code such as `200`, `404`, or `500`.
3. ⏱️ **Response Time** – Calculate how quickly the website responds.
4. ✅ **Status Message** – Display `Online`, `Not Found`, `Server Error`, or other suitable status.
5. 🔗 **URL Validation** – Check whether the entered URL is in a valid format.
6. 🔄 **Multiple Checks** – Allow the user to check more than one website.
7. 📊 **Monitoring Summary** – Display the number of websites checked and their status.
8. ⚠️ **Error Handling** – Handle connection timeout, invalid URL, and network errors.
9. 🚪 **Exit Option** – Allow the user to safely exit the program.

````python
# 🌐 WEBSITE HEALTH MONITOR

import requests
import time
from urllib.parse import urlparse


# Store monitoring history
history = []


# Check whether URL is valid
def is_valid_url(url):
    try:
        result = urlparse(url)

        if result.scheme in ["http", "https"] and result.netloc:
            return True

        return False

    except Exception:
        return False


# Check website health
def check_website(url):

    print("\n" + "=" * 60)
    print("🌐 WEBSITE HEALTH MONITOR")
    print("=" * 60)

    print("Website      :", url)

    # Validate URL
    if not is_valid_url(url):
        print("❌ Invalid URL!")
        return

    try:
        print("🔄 Checking website...")

        start_time = time.time()

        # Send request
        response = requests.get(
            url,
            timeout=10
        )

        end_time = time.time()

        # Calculate response time
        response_time = end_time - start_time

        status_code = response.status_code

        # Determine status
        if 200 <= status_code < 300:
            status = "✅ ONLINE"

        elif 300 <= status_code < 400:
            status = "🔄 REDIRECTED"

        elif 400 <= status_code < 500:
            status = "❌ CLIENT ERROR"

        elif 500 <= status_code < 600:
            status = "⚠️ SERVER ERROR"

        else:
            status = "❓ UNKNOWN"

        # Display result
        print("-" * 60)
        print("Status Code  :", status_code)
        print("Response Time:", f"{response_time:.2f} seconds")
        print("Status       :", status)
        print("-" * 60)

        # Save history
        history.append({
            "website": url,
            "status_code": status_code,
            "response_time": response_time,
            "status": status
        })

    except requests.exceptions.Timeout:
        print("❌ Request timed out!")
        print("Status       : 🔴 OFFLINE / SLOW")

    except requests.exceptions.ConnectionError:
        print("❌ Could not connect to the website!")
        print("Status       : 🔴 OFFLINE")

    except requests.exceptions.RequestException as error:
        print("❌ Request Error:", error)

    except Exception as error:
        print("❌ Unexpected Error:", error)


# Display monitoring history
def show_history():

    print("\n" + "=" * 70)
    print("📊 WEBSITE MONITORING HISTORY")
    print("=" * 70)

    if not history:
        print("No websites checked yet.")
        return

    print(
        f"{'No.':<6}"
        f"{'Website':<30}"
        f"{'Code':<8}"
        f"{'Time':<12}"
        f"{'Status'}"
    )

    print("-" * 70)

    for number, item in enumerate(history, 1):

        print(
            f"{number:<6}"
            f"{item['website'][:28]:<30}"
            f"{item['status_code']:<8}"
            f"{item['response_time']:.2f}s{'':<7}"
            f"{item['status']}"
        )

    print("=" * 70)


# Display summary
def show_summary():

    print("\n" + "=" * 50)
    print("📈 MONITORING SUMMARY")
    print("=" * 50)

    total = len(history)

    if total == 0:
        print("No websites checked yet.")
        return

    online = 0
    errors = 0

    for item in history:

        if 200 <= item["status_code"] < 400:
            online += 1
        else:
            errors += 1

    print("Total Checks :", total)
    print("Online       :", online)
    print("Errors       :", errors)

    print("=" * 50)


# Main program
while True:

    print("\n" + "=" * 55)
    print("🌐 WEBSITE HEALTH MONITOR")
    print("=" * 55)

    print("1. 🔍 Check Website")
    print("2. 📊 View Monitoring History")
    print("3. 📈 View Summary")
    print("4. 🚪 Exit")

    print("=" * 55)

    choice = input("Enter your choice (1-4): ")

    if choice == "1":

        url = input(
            "\nEnter permitted website URL: "
        ).strip()

        check_website(url)

    elif choice == "2":

        show_history()

    elif choice == "3":

        show_summary()

    elif choice == "4":

        print("\n👋 Thank you for using Website Health Monitor!")
        break

    else:

        print("❌ Invalid choice! Please select 1-4.")
````
